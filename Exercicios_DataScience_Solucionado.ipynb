## 2. Exercícios de SciPy

### 1. Integral de f(x) = x² sen(x) no intervalo de 0 a π

```python
def f1(x):
    return x**2 * np.sin(x)

resultado1, erro1 = integrate.quad(f1, 0, np.pi)
print("Resultado da integral:", resultado1)
print("Erro estimado:", erro1)
```

### 2. EDO dy/dx = x - y, y(0) = 1, no intervalo de 0 a 10

```python
def eq2(x, y):
    return x - y

solucao2 = integrate.solve_ivp(eq2, [0, 10], [1], t_eval=np.linspace(0, 10, 50))
print("x:", solucao2.t)
print("y:", solucao2.y[0])
```

### 3. Raízes de f(x) = x⁴ - 3x³ + 2

```python
def f3(x):
    return x**4 - 3*x**3 + 2

# Testando algumas estimativas iniciais diferentes
for chute in [-2, 0, 1, 3]:
    raiz3 = optimize.fsolve(f3, chute)
    print(f"Chute inicial {chute} -> raiz encontrada: {raiz3}")
```

### 4. Maximizar f(x) = -x² + 4x + 1

```python
def f4(x):
    # Para maximizar f(x), minimizamos -f(x)
    return -(-x**2 + 4*x + 1)

resultado4 = optimize.minimize(f4, 0)
x_max4 = resultado4.x[0]
print("x que maximiza a função:", x_max4)
print("Valor máximo:", -resultado4.fun)
```

### 5. Transformada de Fourier de sinal sen(2πt) + cos(4πt)

```python
t5 = np.linspace(0, 1, 500)
sinal5 = np.sin(2 * np.pi * t5) + np.cos(4 * np.pi * t5)
frequencias5 = spfft.fft(sinal5)
print("Transformada de Fourier (primeiros 10 valores):", frequencias5[:10])
```

### 6. Matriz inversa de uma matriz 5x5 de números aleatórios

```python
matriz6_scipy = np.random.rand(5, 5)
inversa6 = linalg.inv(matriz6_scipy)
print("Matriz original:\n", matriz6_scipy)
print("Matriz inversa:\n", inversa6)
```

### 7. Interpolação cúbica dos pontos (0,0), (1,2), (2,3), (3,5), (4,4)

```python
x7 = np.array([0, 1, 2, 3, 4])
y7 = np.array([0, 2, 3, 5, 4])
f7 = interpolate.interp1d(x7, y7, kind='cubic')
x_novos7 = np.linspace(0, 4, 9)
y_novos7 = f7(x_novos7)
print("Valores interpolados:", y_novos7)
```

### 8. Sistema 3x + 2y = 5 e x + 4y = 6

```python
A8 = np.array([[3, 2], [1, 4]])
B8 = np.array([5, 6])
solucao8 = linalg.solve(A8, B8)
print("Solução [x, y]:", solucao8)
```

### 9. Ajuste de curva quadrática aos pontos (1,2),(2,4),(3,5),(4,7),(5,8)

```python
def parabola9(x, a, b, c):
    return a * x**2 + b * x + c

x9 = np.array([1, 2, 3, 4, 5])
y9 = np.array([2, 4, 5, 7, 8])
parametros9, _ = optimize.curve_fit(parabola9, x9, y9)
print("Parâmetros ajustados [a, b, c]:", parametros9)
```

### 10. Integral dupla de f(x, y) = xy + x² em [0,2] para x e [0,1] para y

```python
def f10(y, x):
    return x * y + x**2

resultado10, erro10 = integrate.dblquad(f10, 0, 2, lambda x: 0, lambda x: 1)
print("Resultado da integral dupla:", resultado10)
```

### 11. Autovalores e autovetores de uma matriz 4x4 de inteiros aleatórios

```python
matriz11_scipy = np.random.randint(1, 10, size=(4, 4))
autovalores11, autovetores11 = linalg.eig(matriz11_scipy)
print("Matriz:\n", matriz11_scipy)
print("Autovalores:", autovalores11)
print("Autovetores:\n", autovetores11)
```

### 12. Raiz de e^x - x² = 0 próxima de x = 1

```python
def f12(x):
    return np.exp(x) - x**2

raiz12 = optimize.fsolve(f12, 1)
print("Raiz encontrada:", raiz12)
```

### 13. Determinante de matriz 6x6 de números aleatórios entre -10 e 10

```python
matriz13_scipy = np.random.uniform(-10, 10, size=(6, 6))
determinante13 = linalg.det(matriz13_scipy)
print("Matriz:\n", matriz13_scipy)
print("Determinante:", determinante13)
```

### 14. Minimizar f(x, y) = (x - 2)² + (y - 3)²

```python
def f14(v):
    x, y = v
    return (x - 2)**2 + (y - 3)**2

resultado14 = optimize.minimize(f14, [0, 0])
print("Mínimo encontrado em [x, y]:", resultado14.x)
```

### 15. Transformada de Fourier inversa de um sinal complexo aleatório

```python
sinal15 = np.random.rand(20) + 1j * np.random.rand(20)
sinal15_transformado = spfft.fft(sinal15)
sinal15_reconstruido = spfft.ifft(sinal15_transformado)
print("Sinal original (primeiros 5):", sinal15[:5])
print("Sinal reconstruído (primeiros 5):", sinal15_reconstruido[:5])
```

### 16. EDO de segunda ordem d²y/dx² + y = 0, y(0) = 1, y'(0) = 0

```python
# Convertendo em sistema de 1ª ordem: y0 = y, y1 = y'
def eq16(x, estado):
    y0, y1 = estado
    return [y1, -y0]

solucao16 = integrate.solve_ivp(eq16, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 50))
print("x:", solucao16.t)
print("y:", solucao16.y[0])
```

### 17. Raiz cúbica de 27
