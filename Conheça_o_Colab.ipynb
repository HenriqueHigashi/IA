```
# Primeiros números da sequência
a = 0
b = 1

# Mostra 10 números
for i in range(10):
    print(a)

    # Calcula o próximo número
    proximo = a + b
    a = b
    b = proximo
```

```
# Cria uma lista
numeros = [10, 5, 20, 3, 15]

# Define inicialmente o maior e o menor
maior = numeros[0]
menor = numeros[0]

# Verifica todos os números
for numero in numeros:
    if numero > maior:
        maior = numero

    if numero < menor:
        menor = numero

# Mostra os resultados
print("Maior número:", maior)
print("Menor número:", menor)
```

```
# Cria uma lista
numeros = [10, 8, 7, 9]

soma = 0

# Soma os números
for numero in numeros:
    soma = soma + numero

# Calcula a média
media = soma / len(numeros)

# Mostra o resultado
print("A média é:", media)
```

```
# Pede uma frase
texto = input("Digite uma frase: ")

# Separa as palavras
palavras = texto.split()

# Cria um dicionário vazio
contagem = {}

# Conta as palavras
for palavra in palavras:
    if palavra in contagem:
        contagem[palavra] = contagem[palavra] + 1
    else:
        contagem[palavra] = 1

# Mostra o resultado
print(contagem)
```

```
# Pede um número
numero = int(input("Digite um número: "))

divisores = 0

# Verifica os divisores do número
for i in range(1, numero + 1):
    if numero % i == 0:
        divisores = divisores + 1

# Um número primo possui apenas dois divisores
if divisores == 2:
    print("É um número primo")
else:
    print("Não é um número primo")
```
