```python
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.svm import SVC
```

```python
df = pd.read_csv("bank.csv", sep=";")
print("Shape:", df.shape)
```

```python
print("Dupliacados:", df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop= True)
```

```python
df = df.replace("unknown", np.nan)
print("Faltantes por Coluna:\n", df.isna().sum()[df.isna().sum() > 0])
```

```python
x = df.drop(columns="y")
y = (df["y"] == "yes").astype(int)
```

```python
num_cols = x.select_dtypes(include="number").columns.tolist()
cat_cols = x.select_dtypes(exclude="number").columns.tolist()
```

```python
pre = ColumnTransformer([
    ("num", Pipeline([("imp", SimpleImputer(strategy="median")),
    ("sc", StandardScaler())]), num_cols),
    ("cat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
    ("oh", OneHotEncoder(handle_unknown="ignore"))]), cat_cols),])
pipe = Pipeline([("pre", pre), ("svc", SVC())])
```

```python
param_grid = {
    "svc__C": [0.1,1,10],
    "svc__kernel": ["linear", "rbf"],
    "svc__gamma": ["scale", 0.01],
    "svc__class_weight" : [None,"Balanced"],
}
```

```python
x_tr, x_te, y_tr, y_te = train_test_split(x, y, test_size=0.2, stratify=y, random_state=42)
```

```python
grid = GridSearchCV(pipe, param_grid, cv=5, scoring="f1", n_jobs=1, verbose=1)
grid.fit(x_tr, y_tr)
```

```python
print("\nMelhores parâmetros:", grid.best_params_)
print("Melhor F1 (CV): %.3f" % grid.best_score_)
```

```python
pred = grid.predict(x_te)
print("\nMatriz de confusão: \n", confusion_matrix(y_te, pred))
print("\n", classification_report(y_te, pred, target_names=["no", "yes"]))
```
